```python
data_dir = Path("/Volumes/Misophonia_EEG/") 
subject_list = ["103", "104", "105", "106", "107", "108", "109", "110", "112", "113", "115", "119", "121", "122", "123", "124", "126", "129", "130", "131", "133", "134", "145", "147", "148", "152", "153", "155", "156", "157", "158", "159", "162", "174", "178", "179", "196", "202", "213"]
cutoff_date = datetime(2026, 10, 5) 

for subject in subject_list:
    subject_dir = data_dir / subject
    file = subject_dir / f"{subject}_stage2_cleaned_raw.fif"

    if file.exists():
        file_mod_time = datetime.fromtimestamp(file.stat().st_mtime)
        if file_mod_time > cutoff_date:
            continue
        else:
            print(f"{subject}: file was NOT modified after {cutoff_date}.")
    else:
        print(f"{subject}: file {file} does not exist.")

    
```
